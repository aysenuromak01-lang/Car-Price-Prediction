# Araba Fiyat Tahmini

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score
import joblib
```

```python
df = pd.read_csv("CarPrice.csv")
```

```python
df.info()
```

```python
df
```

```python
# 2. Fiyatı en çok etkileyen kritik sayısal özellikleri seçme
features = ["wheelbase", "carlength", "carwidth", "curbweight", 
            "enginesize", "horsepower", "citympg", "highwaympg"]
```

```python
x = df[features]
y = df["price"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
```

```python
model = DecisionTreeRegressor(random_state=42)
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
print("Model R2 Başarı Skoru:", r2_score(y_test, y_pred))
```

```python
# Modeli Streamlit uygulaması için kaydetme
joblib.dump(model, "car_price_model.pkl")
print("Model başarıyla 'car_price_model.pkl' olarak kaydedildi!")
```
